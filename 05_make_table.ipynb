# 05 — Generate Table I

Formats the actual evaluation results and lists their differences from the published values.


```python
from pathlib import Path
import subprocess
import sys
ROOT = Path.cwd()
assert (ROOT / 'scripts').is_dir(), 'Start from the repository root'
ENCODERS = ['hear']  # Add 'qwen' after choosing its extraction mode in02.
EMBEDDINGS_DIR = ROOT / 'data/embeds'
CLEANING_DIR = ROOT / 'data/cleaning'
RESULTS_DIR = ROOT / 'result/table_i'

def run_step(name, *options):
    subprocess.run([sys.executable, '-u', str(ROOT / 'scripts' / name),
                    *map(str, options)], cwd=ROOT, check=True)
```

```python
run_step('05_make_table.py', '--encoders', *ENCODERS, '--results-dir', RESULTS_DIR)
```
