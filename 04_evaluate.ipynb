# 04 — Evaluate

Reads the generated cleaning decisions and evaluates the five arms with the original grouped folds, classifiers and aggregation.


```python
from pathlib import Path
import subprocess
import sys
ROOT = Path.cwd()
assert (ROOT / 'scripts').is_dir(), 'Start from the repository root'
ENCODERS = ['hear']  # Add 'qwen' after choosing its extraction mode in02.
EMBEDDINGS_DIR = ROOT / 'data/embeds'
CLEANING_DIR = ROOT / 'data/cleaning'
RESULTS_DIR = ROOT / 'result/table_i'

def run_step(name, *options):
    subprocess.run([sys.executable, '-u', str(ROOT / 'scripts' / name),
                    *map(str, options)], cwd=ROOT, check=True)
```

```python
run_step('04_evaluate.py', '--encoders', *ENCODERS, '--embeddings-dir', EMBEDDINGS_DIR, '--cleaning-dir', CLEANING_DIR, '--output-dir', RESULTS_DIR, '--workers', 16)
```
