# Paper Table I study

Run 01 and 02 first. This notebook calls the port of the paper study:
PCA-256 → Fisher projection → KDE posterior → GMM/BIC split per dataset.
TB clips are classified as **Acoustically Consistent TB** or **Acoustically Ambiguous TB**.
Evaluation retains the paper's Original, Random (15 repeats), CL (PBNR),
CL (CL), and Proposed stages, grouped five-fold CV and three classifiers.

Outputs: `result/table_i/table_i.{csv,md}`, comparison against published values,
and original overall/tbcount/LODO CSVs. LODO is supplementary to Table I.
Regenerated embeddings do not guarantee identical paper numbers.
The prior exploratory notebook is preserved under `extras/` and is unvalidated.

For the independent Qwen experiment, use the same `QWEN_MODE="independent"` setting as notebook 02. This changes features/results; output is separate from paper reproduction.



```python
from pathlib import Path
import os
import subprocess
import sys
QWEN_MODE = globals().get("QWEN_MODE", "paper")  # Match notebook 02; set explicitly after a kernel restart
ROOT = Path.cwd()
EMBEDDINGS_DIR = ROOT / ("data/embeds/independent" if QWEN_MODE == "independent" else "data/embeds")
OUTPUT_DIR = ROOT / ("result/independent_table_i" if QWEN_MODE == "independent" else "result/table_i")
assert (ROOT / "paper_study.py").is_file(), "Start from the repository root"
env = os.environ.copy()
for name in ["OMP_NUM_THREADS", "OPENBLAS_NUM_THREADS", "MKL_NUM_THREADS"]:
    env[name] = "1"
subprocess.run([sys.executable, "-u", str(ROOT / "paper_study.py"), "--workers", "16",
                "--embeddings-dir", str(EMBEDDINGS_DIR), "--output-dir", str(OUTPUT_DIR)],
               cwd=ROOT, env=env, check=True)
```
